```python
# Initialize OpenAI client with API key
from openai import OpenAI

api_key = "YOUR_OPENAI_API_KEY"  # Enter your OpenAI API key here
client = OpenAI(api_key=api_key)
```

```python
# View available models with this API key
models = client.models.list()
for model in sorted(m.id for m in models.data):
    print(model)
```

```python
# Define sports domain chatbot with strict guardrails
messages = [
    {
        "role": "system",
        "content": (
            "You are an expert sports chatbot with complete knowledge of all sports, "
            "rules, athletes, leagues, tournaments, and history. Answer only questions "
            "related to sports. Politely decline and abstain from answering questions "
            "about any other domain."
        ),
    }
]

def sports_chatbot(user_input):
    messages.append({"role": "user", "content": user_input})
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages
    )
    reply = response.choices[0].message.content
    messages.append({"role": "assistant", "content": reply})
    return reply
```

```python
# Test queries for domain verification and refusal
print("Sports Query:", sports_chatbot("Who has won the most FIFA World Cups?"))
print("Non-Sports Query:", sports_chatbot("Explain how photosynthesis works."))
```

```python
# Interactive chatbot loop (type 'exit' to quit)
print("Sports Chatbot is active. Ask any sports question (type 'exit' to quit):")
while True:
    user_query = input("\nYou: ")
    if user_query.lower().strip() in ["exit", "quit"]:
        print("Chatbot: Goodbye!")
        break
    if user_query.strip():
        print(f"Chatbot: {sports_chatbot(user_query)}")
```
